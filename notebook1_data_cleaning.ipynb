```python
# Εισαγωγή βιβλιοθηκών
import pandas as pd      # Για επεξεργασία πινάκων δεδομένων
import numpy as np       # Για αριθμητικές πράξεις
```

```python
# Φόρτωμα αρχείου από τον υπολογιστή μέσω Google Colab
from google.colab import files
uploaded = files.upload()
# --> Διαλέξτε το αρχείο: BoG_PayStatsAnnual_EL_2025-07-30.xlsx
```

```python
# Ανάγνωση του Excel αρχείου σε DataFrame
df_raw = pd.read_excel('BoG_PayStatsAnnual_EL_2025-07-30.xlsx')

# Πρώτη εξερεύνηση: διαστάσεις και χρονικό εύρος
print(f'Διαστάσεις dataset: {df_raw.shape[0]} γραμμές x {df_raw.shape[1]} στήλες')
print(f'Χρονικό εύρος: {df_raw["Έτος_αναφοράς"].min()} – {df_raw["Έτος_αναφοράς"].max()}')
df_raw.head()
```

```python
# Επιλογή μόνο των στηλών που χρειαζόμαστε για την ανάλυση
df_clean = df_raw[[
    'Έτος_αναφοράς',
    'Κάρτες_έκδοσης_εγχώριων_ΠΥΠ_Σύνολο_καρτών_Καθαρός_αριθμός',
    'Τερματικά_εγχώριων_ΠΥΠ_POS_Εγχώρια_Καθαρός_αριθμός',
    'Σύνολο_συναλλαγών_Σύνολο_υπηρεσιών_πληρωμών_Αποσταλείσες_Εγχώριες_Εκατομμύρια',
    'Σύνολο_συναλλαγών_Σύνολο_υπηρεσιών_πληρωμών_Αποσταλείσες_Εγχώριες_Εκατομμύρια_ευρώ',
    'Σύνολο_συναλλαγών_Υπηρεσίες_πληρωμών_Αποσταλείσες_Μεταφορές_πιστώσεων_Εγχώριες_Εκατομμύρια',
    'Σύνολο_συναλλαγών_Υπηρεσίες_πληρωμών_Αποσταλείσες_Άμεσες_χρεώσεις_Εγχώριες_Εκατομμύρια',
    'Σύνολο_συναλλαγών_Υπηρεσίες_πληρωμών_Αποσταλείσες_Επιταγές_Εγχώριες_Εκατομμύρια',
]].copy()

# Μετονομασία στηλών για ευκολία
df_clean.columns = [
    'Έτος',
    'Αριθμός_Καρτών',
    'POS_Τερματικά',
    'Συναλλαγές_Εκατ',
    'Αξία_Εκατ_Ευρώ',
    'Μεταφορές_Πιστώσεων_Εκατ',
    'Άμεσες_Χρεώσεις_Εκατ',
    'Επιταγές_Εκατ'
]

df_clean.head()
```

```python
# Έλεγχος ποιότητας δεδομένων: κενές τιμές, διπλότυπα, τύποι
print('=== ΕΛΕΓΧΟΣ ΠΟΙΟΤΗΤΑΣ ΔΕΔΟΜΕΝΩΝ ===')
print('\n1. Κενές τιμές ανά στήλη:')
print(df_clean.isnull().sum())
print(f'\n2. Διπλότυπες γραμμές: {df_clean.duplicated().sum()}')
print('\n3. Τύποι δεδομένων:')
print(df_clean.dtypes)
```

```python
# Διόρθωση τύπων δεδομένων και ταξινόμηση χρονολογικά
df_clean['Έτος'] = df_clean['Έτος'].astype(int)
df_clean = df_clean.sort_values('Έτος').reset_index(drop=True)

print('✅ Τελικό καθαρό dataset:')
print(f'Γραμμές: {df_clean.shape[0]} | Στήλες: {df_clean.shape[1]}')
print(f'Χρονικό εύρος: {df_clean["Έτος"].min()} – {df_clean["Έτος"].max()}')
df_clean
```

```python
# Εξαγωγή καθαρού dataset σε Excel και CSV
# Το Excel θα χρησιμοποιηθεί για το Power BI dashboard
df_clean.to_excel('bog_payments_clean.xlsx', index=False)
df_clean.to_csv('bog_payments_clean.csv', index=False)

# Κατέβασμα αρχείων στον υπολογιστή
files.download('bog_payments_clean.xlsx')
files.download('bog_payments_clean.csv')

print('✅ Αρχεία αποθηκεύτηκαν επιτυχώς!')
print('   bog_payments_clean.xlsx --> για Power BI')
print('   bog_payments_clean.csv  --> για Notebook 2')
```
